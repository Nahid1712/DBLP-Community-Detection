```python
import gzip
import networkx as nx
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import defaultdict
from lxml import etree
import community as community_louvain
import re
import html
import os
import warnings
warnings.filterwarnings("ignore")


class AdaptiveDBLPAnalyzer:

    def __init__(self, path):
        self.path = path
        self.G_year = {}
        self.partition_year = {}
        self.mod_year = {}

        # default controls
        self.resolution = 1.0

        self.target_low = 0.3
        self.target_high = 0.8

    # ---------------- CLEANING ----------------
    def clean(self, text):
        if text is None:
            return ""
        text = html.unescape(text)
        return " ".join(text.split())

    def clean_author(self, name):
        name = self.clean(name)
        name = re.sub(r"\s+\d+$", "", name)
        return name.strip()

    # ---------------- DBLP STREAM ----------------
    def stream(self):
        with gzip.open(self.path, "rt", encoding="iso-8859-1", errors="ignore") as f:
            buf = []
            inside = False

            for line in f:
                line = line.strip()

                if line.startswith("<article") or line.startswith("<inproceedings"):
                    buf = [line]
                    inside = True

                elif inside:
                    buf.append(line)

                    if line.endswith("</article>") or line.endswith("</inproceedings>"):
                        xml = "\n".join(buf)
                        buf = []
                        inside = False

                        try:
                            root = etree.fromstring(xml.encode("iso-8859-1"))
                            yield root
                        except:
                            continue

    # ---------------- BUILD GRAPH ----------------
    def build(self, start=2000, end=2020, max_papers=50000):

        yearly_edges = defaultdict(lambda: defaultdict(int))
        count = 0

        for paper in self.stream():
            year = paper.findtext("year")
            if not year:
                continue

            try:
                year = int(year)
            except:
                continue

            if year < start or year > end:
                continue

            authors = [self.clean_author(a.text) for a in paper.findall("author") if a.text]
            authors = [a for a in authors if len(a) > 1]

            if len(authors) < 2:
                continue

            for i in range(len(authors)):
                for j in range(i + 1, len(authors)):
                    a, b = sorted([authors[i], authors[j]])
                    yearly_edges[year][(a, b)] += 1

            count += 1
            if count >= max_papers:
                break

        # build graphs
        for y, edges in yearly_edges.items():
            G = nx.Graph()
            for (a, b), w in edges.items():
                G.add_edge(a, b, weight=w)

            G.remove_nodes_from(list(nx.isolates(G)))
            self.G_year[y] = G

            print(f"Year {y}: nodes={G.number_of_nodes()} edges={G.number_of_edges()}")

    # ---------------- MODULARITY ----------------
    def modularity(self, G, partition):
        m = G.size(weight="weight")
        if m == 0:
            return 0

        comms = defaultdict(list)
        for n, c in partition.items():
            comms[c].append(n)

        Q = 0
        for nodes in comms.values():
            if not nodes:
                continue

            internal = 0
            degree_sum = 0
            S = set(nodes)

            for n in nodes:
                degree_sum += G.degree(n, weight="weight")
                for nb, d in G[n].items():
                    if nb in S:
                        internal += d.get("weight", 1)

            internal /= 2

            Q += (internal / (2 * m)) - (degree_sum / (2 * m)) ** 2

        return Q

    # ---------------- GRAPH CONTROL ----------------
    def densify(self, G):
        """
        LOWERS modularity by adding weak cross-community links
        """
        nodes = list(G.nodes())
        added = 0

        for i in range(len(nodes)):
            for j in range(i + 1, len(nodes)):
                if not G.has_edge(nodes[i], nodes[j]):
                    if np.random.rand() < 0.0005:  # controlled noise
                        G.add_edge(nodes[i], nodes[j], weight=0.05)
                        added += 1

        print(f"   🔵 densified edges added: {added}")
        return G

    def sharpen(self, G):
        """
        INCREASES modularity by removing weak edges
        """
        to_remove = []

        for u, v, d in G.edges(data=True):
            if d.get("weight", 1) <= 1:
                if np.random.rand() < 0.3:
                    to_remove.append((u, v))

        G.remove_edges_from(to_remove)
        print(f"   🔴 sharpened edges removed: {len(to_remove)}")
        return G

    # ---------------- LOUVAIN ----------------
    def run_louvain(self, G):
        return community_louvain.best_partition(
            G,
            weight="weight",
            resolution=self.resolution
        )

    # ---------------- CONTROL LOOP ----------------
    def optimize_year(self, year, max_iter=6):

        G = self.G_year[year]

        for it in range(max_iter):

            partition = self.run_louvain(G)
            Q = self.modularity(G, partition)

            print(f"Year {year} | iter {it} | Q={Q:.4f} | res={self.resolution:.2f}")

            if self.target_low <= Q <= self.target_high:
                print("   ✅ TARGET ACHIEVED")
                self.partition_year[year] = partition
                self.mod_year[year] = Q
                return

            # HIGH MODULARITY → densify
            if Q > self.target_high:
                self.resolution *= 0.7
                G = self.densify(G)

            # LOW MODULARITY → sharpen
            elif Q < self.target_low:
                self.resolution *= 1.3
                G = self.sharpen(G)

        self.partition_year[year] = partition
        self.mod_year[year] = Q

    # ---------------- FULL PIPELINE ----------------
    def run(self, start=2000, end=2020):

        self.build(start, end)

        for year in sorted(self.G_year.keys()):
            print("\n==============================")
            print("Processing year:", year)
            print("==============================")
            self.optimize_year(year)

        self.report()

    # ---------------- REPORT ----------------
    def report(self):
        print("\n\n=========== FINAL REPORT ===========")

        for y in sorted(self.mod_year):
            q = self.mod_year[y]
            status = "GOOD" if 0.3 <= q <= 0.8 else "OUT"
            print(f"{y}: Q={q:.4f} -> {status}")


# ---------------- RUN PIPELINE ----------------

path = r"C:\Users\Nahid\Downloads\dblp.xml.gz"

model = AdaptiveDBLPAnalyzer(path)
model.run(start=2000, end=2020)
```

```python
import matplotlib.pyplot as plt
import networkx as nx
import random

def visualize_year(G, partition, year, top_n=50):
    """
    Visualize top_n nodes by degree with community colors
    """
    # Sort nodes by degree
    degrees = dict(G.degree())
    top_nodes = sorted(degrees, key=degrees.get, reverse=True)[:top_n]

    H = G.subgraph(top_nodes).copy()

    # Community colors
    communities = list(set(partition.values()))
    color_map = {}
    for i, c in enumerate(communities):
        color_map[c] = plt.cm.tab20(i % 20)

    node_colors = [color_map.get(partition.get(n, -1), (0.5,0.5,0.5)) for n in H.nodes()]

    plt.figure(figsize=(12, 10))
    pos = nx.spring_layout(H, seed=42, k=0.5)

    nx.draw_networkx_nodes(H, pos, node_color=node_colors, node_size=[100 + degrees[n]*20 for n in H.nodes()])
    nx.draw_networkx_edges(H, pos, alpha=0.4)
    nx.draw_networkx_labels(H, pos, font_size=8)

    Q = model.mod_year.get(year, 0)
    plt.title(f"DBLP Year {year} | Modularity Q={Q:.3f} | Top {top_n} nodes", fontsize=16)
    plt.axis('off')
    plt.show()
```

```python
# Visualize a specific year, e.g., 2010
year_to_plot = 2012
G = model.G_year[year_to_plot]
partition = model.partition_year[year_to_plot]

visualize_year(G, partition, year_to_plot)
```

```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np

def plot_dblp_report(model, top_n=60, cols=3):
    """
    Full report-style visualization for all years
    """

    years = sorted(model.G_year.keys())
    rows = int(np.ceil(len(years) / cols))

    plt.figure(figsize=(cols * 6, rows * 5))

    for i, year in enumerate(years):

        G = model.G_year[year]
        partition = model.partition_year.get(year, {})
        Q = model.mod_year.get(year, 0)

        if G is None or len(G) == 0 or not partition:
            continue

        # --- Top nodes by degree ---
        deg = dict(G.degree())
        top_nodes = sorted(deg, key=deg.get, reverse=True)[:top_n]
        H = G.subgraph(top_nodes).copy()

        # --- colors by community ---
        colors = [partition.get(n, -1) for n in H.nodes()]

        pos = nx.spring_layout(H, seed=42, k=0.6)

        ax = plt.subplot(rows, cols, i + 1)

        nx.draw_networkx_nodes(
            H, pos,
            node_color=colors,
            cmap=plt.cm.tab20,
            node_size=[50 + deg[n] * 20 for n in H.nodes()],
            alpha=0.9,
            ax=ax
        )

        nx.draw_networkx_edges(H, pos, alpha=0.25, ax=ax)

        nx.draw_networkx_labels(H, pos, font_size=7, ax=ax)

        # --- Title with modularity ---
        status = "GOOD" if 0.3 <= Q <= 0.8 else "OUT"
        ax.set_title(f"{year} | Q={Q:.3f} | {status}", fontsize=11)

        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
plot_dblp_report(model, top_n=60, cols=3)
```

```python
import matplotlib.pyplot as plt

def plot_modularity_trend(model):
    years = sorted(model.mod_year.keys())
    values = [model.mod_year[y] for y in years]

    plt.figure(figsize=(10,5))
    plt.plot(years, values, marker='o', linewidth=2)

    # target band
    plt.axhline(0.3, color='green', linestyle='--', label='Lower bound (0.3)')
    plt.axhline(0.8, color='red', linestyle='--', label='Upper bound (0.8)')
    plt.fill_between(years, 0.3, 0.8, color='green', alpha=0.1)

    plt.title("DBLP Modularity Trend Over Time")
    plt.xlabel("Year")
    plt.ylabel("Modularity (Q)")
    plt.ylim(0, 1)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()
```

```python
import numpy as np
import pandas as pd

def community_tracking(model):
    years = sorted(model.partition_year.keys())

    def get_comms(part):
        comms = {}
        for node, c in part.items():
            comms.setdefault(c, set()).add(node)
        return list(comms.values())

    def jaccard(a, b):
        return len(a & b) / len(a | b)

    results = []

    for i in range(len(years) - 1):
        y1, y2 = years[i], years[i+1]

        p1 = model.partition_year[y1]
        p2 = model.partition_year[y2]

        c1 = get_comms(p1)
        c2 = get_comms(p2)

        # compute best matching similarity
        sims = []
        for A in c1:
            best = 0
            for B in c2:
                best = max(best, jaccard(A, B))
            sims.append(best)

        stability = np.mean(sims) if sims else 0

        results.append({
            "year_pair": f"{y1}-{y2}",
            "stability": stability
        })

    df = pd.DataFrame(results)

    plt.figure(figsize=(10,5))
    plt.plot(df["year_pair"], df["stability"], marker='o')

    plt.xticks(rotation=45)
    plt.ylim(0,1)
    plt.title("Community Stability Across Years (Jaccard Similarity)")
    plt.ylabel("Stability")
    plt.grid(alpha=0.3)
    plt.show()

    return df
```
